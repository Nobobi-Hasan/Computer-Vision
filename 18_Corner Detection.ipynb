```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
img_flat_chess = cv2.imread("DATA/flat_chessboard.png")
rgb_flat_chess = cv2.cvtColor(img_flat_chess, cv2.COLOR_BGR2RGB)

plt.imshow(rgb_flat_chess)
```

```python
gray_flat_chess = cv2.cvtColor(img_flat_chess, cv2.COLOR_BGR2GRAY)
plt.imshow(gray_flat_chess, cmap='gray')
```

```python
img_real_chess = cv2.imread("DATA/real_chessboard.jpg")
rgb_real_chess = cv2.cvtColor(img_real_chess, cv2.COLOR_BGR2RGB)

plt.imshow(rgb_real_chess)
```

```python
gray_real_chess = cv2.cvtColor(img_real_chess, cv2.COLOR_BGR2GRAY)
plt.imshow(gray_real_chess, cmap='gray')
```

# Harris Corner Detection

## On Flat Chess

```python
# At 1st, convert to float
gray_flat_chess = np.float32(gray_flat_chess)

dst = cv2.cornerHarris(src=gray_flat_chess, blockSize=2, ksize=3, k=0.04)
plt.imshow(dst, cmap='gray')
```

```python
dst = cv2.dilate(dst, None)  # dilate for visualization improvement
plt.imshow(dst, cmap='gray')
```

```python
rgb_flat_chess[dst>0.1*dst.max()] = [255,0,0]
plt.imshow(rgb_flat_chess)
```

```python
dst>0.01*dst.max()
```

## On Real Chess

```python
# At 1st, convert to float
gray_real_chess = np.float32(gray_real_chess)

dst = cv2.cornerHarris(src=gray_real_chess, blockSize=2, ksize=3, k=0.04)
plt.imshow(dst, cmap='gray')
```

```python
dst = cv2.dilate(dst, None)  # dilate for visualization improvement
plt.imshow(dst, cmap='gray')
```

```python
rgb_real_chess[dst>0.1*dst.max()] = [255,0,0]
plt.imshow(rgb_real_chess)
```
