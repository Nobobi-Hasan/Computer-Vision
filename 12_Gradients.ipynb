```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
def show_img(img):
    fig = plt.figure(figsize=(12,10))
    ax  = fig.add_subplot(111) #(row, col, subplot)
    ax.imshow(img, cmap='gray')
```

```python
img = cv2.imread("DATA/sudoku.jpg", 0)
show_img(img)
```

# Sobel

```python
sobel_x = cv2.Sobel(img, cv2.CV_64F, 1, 0, ksize=9)
show_img(sobel_x)
```

```python
sobel_y = cv2.Sobel(img, cv2.CV_64F, 0, 1, ksize=9)
show_img(sobel_y)
```

```python
blended = cv2.addWeighted(src1=sobel_x, alpha=0.5, src2=sobel_y, beta=0.5, gamma=0)
show_img(blended)
```

# Laplasian Gradients

```python
laplacian = cv2.Laplacian(img, cv2.CV_64F, ksize=9)
show_img(laplacian)
```

```python
kernel = np.ones(shape=(5,5), dtype=np.uint8)
```

```python
m_grd = cv2.morphologyEx(blended, cv2.MORPH_GRADIENT, kernel)
show_img(m_grd)
```
