```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
img = cv2.imread('DATA/internal_external.png', 0)
plt.imshow(img, cmap='gray')
```

```python
img.shape
```

```python
image, contours, hierarchy  = cv2.findContours(img, mode=cv2.RETR_CCOMP, method=cv2.CHAIN_APPROX_SIMPLE)
```

```python
print(type(contours))
print(len(contours))
```

```python
print(type(hierarchy))
print(len(hierarchy))
print(hierarchy.shape)
```

```python
plt.imshow(image)
```

```python
hierarchy # in hierarchy[0][i][3]  -> -1 = External Contours, 0 and 4 = Internal Contours
```

```python
img_ext_contours = np.zeros(image.shape)

for i in range(len(contours)):
    
    # External Contours
    if hierarchy[0][i][3] == -1:  # -1 = external contour
        cv2.drawContours(img_ext_contours, contours, i, 255, 3)
        
plt.imshow(img_ext_contours, cmap='gray')
```

```python
img_int_contours = np.zeros(image.shape)

for i in range(len(contours)):
    
    # Internal Contours
    if hierarchy[0][i][3] != -1:  # not -1 = internal contour
        cv2.drawContours(img_int_contours, contours, i, 255, 3)
        
plt.imshow(img_int_contours, cmap='gray')
```

```python
img_int1_contours = np.zeros(image.shape)

for i in range(len(contours)):
    
    # Internal Contours
    if hierarchy[0][i][3] == 0:  # not -1 = internal contour
        cv2.drawContours(img_int1_contours, contours, i, 255, 3)
        
plt.imshow(img_int1_contours, cmap='gray')
```

```python
img_int2_contours = np.zeros(image.shape)

for i in range(len(contours)):
    
    # Internal Contours
    if hierarchy[0][i][3] == 4:  # not -1 = internal contour
        cv2.drawContours(img_int2_contours, contours, i, 255, 3)
        
plt.imshow(img_int2_contours, cmap='gray')
```

```python
img_all_contours = np.zeros(image.shape)

for i in range(len(contours)):
    
    # All Contours
    cv2.drawContours(img_all_contours, contours, i, 255, 3)
        
plt.imshow(img_all_contours, cmap='gray')
```
