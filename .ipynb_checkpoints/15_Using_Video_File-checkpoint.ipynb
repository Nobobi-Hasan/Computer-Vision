```python
import cv2
import time

cap = cv2.VideoCapture('DATA/SavedData/TestVideo.mp4')

if cap.isOpened() == False:
    print('ERROR! File Not Found!! OR Wrong Codec Used!!')
 
fps = cap.get(cv2.CAP_PROP_FPS)

while cap.isOpened():
    ret, frame = cap.read()
    
    if ret == True:
        time.sleep(1/fps) # As the TestVideo was recorded with 25 FPS
        cv2.imshow('TestVideo', frame)
    
        if cv2.waitKey(1) & 0xFF == ord('q'):
           break
        
    else:
        break
        
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
duration = frames/fps
print(f"Width: {width}, Height: {height}, Total Frames: {frames}, Duration (s): {duration:.2f}, FPS: {fps}")
        
cap.release()
cv2.destroyAllWindows()
```
