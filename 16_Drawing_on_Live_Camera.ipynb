# Drawing static rectangle on video

```python
import cv2

cap = cv2.VideoCapture(0)

width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

# Top-Left Coordinate of the Rectangle
x1 = width // 2
y1 = height // 2

# Width and Height of the Rectangle
w = width // 4
h = height // 4

# Bottom-Right Coordinate of the Rectangle
x2 = x1 + w
y2 = y1 + h

while True:
    ret, frame = cap.read()
    
    cv2.rectangle(frame, (x1, y1), (x2, y2), (0, 255, 255), 2)
    
    cv2.imshow('Video Testing', frame)
    
    if cv2.waitKey(1) & 0xFF == ord('q'):
       break
    
cap.release()
cv2.destroyAllWindows()
```

# Interactive Drawing

```python
import cv2

cap = cv2.VideoCapture(0)

width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

# Global Variables
pt1 = (0,0)
pt2 = (0,0)
topLeftClicked = False
botRightClicked = False


def draw(event,x,y,flags,param):
    global pt1, pt2, topLeftClicked, botRightClicked
    
    if event == cv2.EVENT_LBUTTONDOWN:
        
            
        # draw circle
        if topLeftClicked == False:
            pt1 = (x,y)
            topLeftClicked = True
            
        # draw circle
        elif botRightClicked == False:
            pt2 = (x,y)
            botRightClicked = True
            
        # reset drawings
        elif topLeftClicked == True and botRightClicked == True:
            pt1 = (0,0)
            pt2 = (0,0)
            topLeftClicked = False
            botRightClicked = False
        
        
cv2.namedWindow(winname="Video Testing")
cv2.setMouseCallback("Video Testing", draw)

while True:
    ret, frame = cap.read()
    
    if topLeftClicked == True and botRightClicked == False:
        cv2.circle(frame, pt1, 20, (0,255,255), -1)
        
    if topLeftClicked == True and botRightClicked == True:
        cv2.rectangle(frame, pt1, pt2, (0,255,255), 2)
    
    cv2.imshow('Video Testing', frame)
    
    if cv2.waitKey(1) & 0xFF == ord('q'):
       break
    
cap.release()
cv2.destroyAllWindows()
```
